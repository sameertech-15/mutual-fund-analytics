```python
top_aum = aum.groupby("fund_house")["aum_crore"].max().sort_values(ascending=False)

top_aum.head(10)
```

```python
top_aum = aum.groupby("fund_house")["aum_crore"].max()

top_aum = top_aum.sort_values(ascending=False)

top_aum.head(10)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10,5))

top_aum.head(10).plot(kind="bar")

plt.title("Top 10 Fund Houses by AUM")

plt.ylabel("AUM Crore")

plt.show()
```

### Insight

SBI Mutual Fund has the highest Assets Under Management (AUM).

ICICI Prudential and HDFC Mutual Fund are the next largest fund houses.

The top fund houses dominate the mutual fund industry.

```python
sip["month"] = pd.to_datetime(sip["month"])

sip.head()
```

```python
plt.figure(figsize=(12,5))

plt.plot(
    sip["month"],
    sip["sip_inflow_crore"]
)

plt.title("Monthly SIP Inflow Trend")

plt.xlabel("Month")

plt.ylabel("SIP Inflow (Crore)")

plt.grid(True)

plt.show()
```

### Insight

SIP inflows show a strong upward trend over the years.

This indicates growing participation of retail investors in mutual funds.

The mutual fund industry continues to attract new investors through SIP investments.

```python
plt.figure(figsize=(12,5))

plt.plot(
    sip["month"],
    sip["active_sip_accounts_crore"]
)

plt.title("Active SIP Accounts Growth")

plt.xlabel("Month")

plt.ylabel("Active SIP Accounts (Crore)")

plt.grid(True)

plt.show()
```

### Insight

The number of active SIP accounts has increased steadily over time.

This indicates growing retail participation in mutual fund investments.

More investors are adopting SIPs as a long-term wealth creation strategy.

```python
category_summary = category.groupby("category")["net_inflow_crore"].sum()

category_summary = category_summary.sort_values(ascending=False)

category_summary.head(10)
```

```python
plt.figure(figsize=(12,6))

category_summary.plot(kind="bar")

plt.title("Category Wise Net Inflows")

plt.ylabel("Net Inflow (Crore)")

plt.xticks(rotation=45)

plt.show()
```

### Insight

Liquid funds attracted the highest inflows among all categories.

Sectoral/Thematic funds also received significant investor interest.

Investors preferred diversified and liquid investment options.

```python
folios["month"] = pd.to_datetime(folios["month"])

plt.figure(figsize=(12,5))

plt.plot(
    folios["month"],
    folios["total_folios_crore"]
)

plt.title("Total Mutual Fund Folios Growth")

plt.xlabel("Month")

plt.ylabel("Total Folios (Crore)")

plt.grid(True)

plt.show()
```

### Insight

The total number of mutual fund folios has increased consistently.

This indicates a growing investor base in the mutual fund industry.

The increasing folio count reflects greater awareness and adoption of mutual fund investments.

```python
top_funds = performance.sort_values(
    by="return_5yr_pct",
    ascending=False
)

top_funds[
    ["scheme_name", "return_5yr_pct"]
].head(10)
```

```python
plt.figure(figsize=(12,6))

plt.bar(
    top_funds["scheme_name"].head(10),
    top_funds["return_5yr_pct"].head(10)
)

plt.title("Top 10 Funds by 5-Year Return")

plt.xlabel("Scheme Name")

plt.ylabel("5-Year Return (%)")

plt.xticks(rotation=90)

plt.tight_layout()

plt.show()
```

### Insight

The top-performing schemes delivered significantly higher 5-year returns.

These funds may be attractive for long-term investors seeking capital appreciation.

Past performance does not guarantee future returns, but it provides useful insight into historical fund performance.

```python
plt.figure(figsize=(10,6))

plt.scatter(
    performance["std_dev_ann_pct"],
    performance["return_5yr_pct"]
)

plt.title("Risk vs Return Analysis")

plt.xlabel("Risk (Standard Deviation %)")

plt.ylabel("5-Year Return (%)")

plt.grid(True)

plt.show()
```

### Insight

Funds with higher risk generally tend to provide higher returns.

However, some high-risk funds do not always deliver proportionally higher returns.

Investors should balance risk and return according to their investment goals.

```python
top_sharpe = performance.sort_values(
    by="sharpe_ratio",
    ascending=False
)

top_sharpe[
    ["scheme_name", "sharpe_ratio"]
].head(10)
```

```python
plt.figure(figsize=(12,6))

plt.bar(
    top_sharpe["scheme_name"].head(10),
    top_sharpe["sharpe_ratio"].head(10)
)

plt.title("Top 10 Funds by Sharpe Ratio")

plt.ylabel("Sharpe Ratio")

plt.xticks(rotation=90)

plt.tight_layout()

plt.show()
```

### Insight

Funds with higher Sharpe Ratios provide better risk-adjusted returns.

These schemes efficiently compensate investors for the risks undertaken.

```python
plt.figure(figsize=(10,6))

plt.hist(
    performance["expense_ratio_pct"],
    bins=10
)

plt.title("Expense Ratio Distribution")

plt.xlabel("Expense Ratio (%)")

plt.ylabel("Number of Funds")

plt.show()
```

### Insight

Most schemes have expense ratios within a moderate range.

Expense ratios directly impact investor returns and should be considered when selecting funds.

```python
numeric_cols = [
    "return_1yr_pct",
    "return_3yr_pct",
    "return_5yr_pct",
    "alpha",
    "beta",
    "sharpe_ratio",
    "expense_ratio_pct",
    "aum_crore"
]

corr = performance[numeric_cols].corr()

corr
```

```python
plt.figure(figsize=(10,8))

plt.imshow(corr)

plt.colorbar()

plt.xticks(
    range(len(corr.columns)),
    corr.columns,
    rotation=90
)

plt.yticks(
    range(len(corr.columns)),
    corr.columns
)

plt.title("Correlation Matrix")

plt.tight_layout()

plt.show()
```

### Insight

The correlation matrix helps identify relationships between performance metrics.

Strong positive correlations indicate metrics moving together, while negative correlations indicate inverse relationships.
