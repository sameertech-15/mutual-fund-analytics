```python
top_funds = performance.sort_values(
    by="return_5yr_pct",
    ascending=False
)

top_funds[
    ["scheme_name", "return_5yr_pct"]
].head(10)
```

```python
plt.figure(figsize=(12,6))

plt.bar(
    top_funds["scheme_name"].head(10),
    top_funds["return_5yr_pct"].head(10)
)

plt.title("Top 10 Funds by 5-Year Return")

plt.xlabel("Scheme Name")

plt.ylabel("5-Year Return (%)")

plt.xticks(rotation=90)

plt.tight_layout()

plt.savefig(
    "../reports/charts/top_10_funds_by_5year_return.png",
    bbox_inches="tight"
)

plt.show()
```

### Insight

The top-performing schemes delivered significantly higher 5-year returns.

These funds may be attractive for long-term investors seeking capital appreciation.

Past performance does not guarantee future returns, but it provides useful insight into historical fund performance.

```python
plt.figure(figsize=(10,6))

plt.scatter(
    performance["std_dev_ann_pct"],
    performance["return_5yr_pct"]
)

plt.title("Risk vs Return Analysis")

plt.xlabel("Risk (Standard Deviation %)")

plt.ylabel("5-Year Return (%)")

plt.grid(True)

plt.savefig(
    "../reports/charts/risk_vs_retun_analysis.png",
    bbox_inches="tight"
)

plt.show()
```

### Insight

Funds with higher risk generally tend to provide higher returns.

However, some high-risk funds do not always deliver proportionally higher returns.

Investors should balance risk and return according to their investment goals.

```python
numeric_cols = [
    "return_1yr_pct",
    "return_3yr_pct",
    "return_5yr_pct",
    "alpha",
    "beta",
    "sharpe_ratio",
    "expense_ratio_pct",
    "aum_crore"
]

corr = performance[numeric_cols].corr()

corr
```

```python
plt.figure(figsize=(10,8))

plt.imshow(corr)

plt.colorbar()

plt.xticks(
    range(len(corr.columns)),
    corr.columns,
    rotation=90
)

plt.yticks(
    range(len(corr.columns)),
    corr.columns
)

plt.title("Correlation Matrix")

plt.tight_layout()

plt.savefig(
    "../reports/charts/correlation_matrix.png",
    bbox_inches="tight"
)

plt.show()
```

### Insight

The correlation matrix helps identify relationships between performance metrics.

Strong positive correlations indicate metrics moving together, while negative correlations indicate inverse relationships.

```python
import pandas as pd
import matplotlib.pyplot as plt

nav = pd.read_csv("../data/raw/02_nav_history.csv")

nav.head()
```

```python
print(nav.columns)
```

```python
nav["date"] = pd.to_datetime(nav["date"])

nav.head()
```

```python
nav_trend = nav.groupby("date")["nav"].mean().reset_index()

plt.figure(figsize=(12,6))

plt.plot(
    nav_trend["date"],
    nav_trend["nav"]
)

plt.title("Average NAV Trend")
plt.xlabel("Date")
plt.ylabel("NAV")
plt.grid(True)

plt.savefig(
    "../reports/charts/average_NAV_trend.png",
    bbox_inches="tight"
)

plt.show()
```

### Insight

NAV values show the overall growth trend of mutual fund schemes.

Temporary corrections and recoveries can be observed during different market phases.

The long-term trend remains positive.

# Category-wise Inflow Heatmap

```python
import seaborn as sns
import matplotlib.pyplot as plt

heatmap_data = category.pivot(
    index="category",
    columns="month",
    values="net_inflow_crore"
)

plt.figure(figsize=(12,6))

sns.heatmap(
    heatmap_data,
    cmap="YlGnBu",
    annot=True,
    fmt=".0f"
)

plt.title("Category-wise Net Inflow Heatmap")
plt.savefig(
    "../reports/charts/categor_wise_net_inflow_heatmap.png",
    bbox_inches="tight"
)
plt.show()
```

### Insight

Some fund categories attract significantly higher inflows than others.

The heatmap highlights periods of strong investor interest across categories.

# Investor Demographics

```python
age_data = investor["age_group"].value_counts()

plt.figure(figsize=(8,8))

plt.pie(
    age_data,
    labels=age_data.index,
    autopct="%1.1f%%"
)

plt.title("Investor Age Group Distribution")
plt.savefig(
    "../reports/charts/investor_age_group_distribution.png",
    bbox_inches="tight"
)
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    x="age_group",
    y="amount_inr",
    data=investor
)

plt.title("Investment Amount by Age Group")
plt.xticks(rotation=45)
plt.savefig(
    "../reports/charts/investment_amount_by_age_group.png",
    bbox_inches="tight"
)

plt.show()
```

```python
state_data = investor.groupby("state")["amount_inr"].sum()

state_data = state_data.sort_values(
    ascending=False
).head(10)

plt.figure(figsize=(10,6))

state_data.plot(kind="barh")

plt.title("Top States by Investment Amount")

plt.xlabel("Investment Amount (INR)")

plt.savefig(
    "../reports/charts/top_states_by_investment_amount.png",
    bbox_inches="tight"
)

plt.show()
```

```python
city_data = investor["city_tier"].value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    city_data,
    labels=city_data.index,
    autopct="%1.1f%%"
)

plt.title("T30 vs B30 Distribution")

plt.savefig(
    "../reports/charts/T30_vs_B30_distribution.png",
    bbox_inches="tight"
)

plt.show()
```
