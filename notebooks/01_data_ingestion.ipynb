# Mutual Fund Analytics Platform
## Data Ingestion Notebook

This notebook loads and validates all raw datasets used in the Mutual Fund Analytics Platform project.

Objectives:
- Load all raw CSV files
- Verify dataset shapes and columns
- Perform basic validation checks
- Prepare data for cleaning and transformation  

```python
import pandas as pd
import os

raw_path = "../data/raw"
files = os.listdir(raw_path)

print("Datasets Found:")
for f in files:
    print(f)
```

```python
fund_master = pd.read_csv("../data/raw/01_fund_master.csv")
nav_history = pd.read_csv("../data/raw/02_nav_history.csv")
aum = pd.read_csv("../data/raw/03_aum_by_fund_house.csv")
sip = pd.read_csv("../data/raw/04_monthly_sip_inflows.csv")

print("Fund Master:", fund_master.shape)
print("NAV History:", nav_history.shape)
print("AUM:", aum.shape)
print("SIP:", sip.shape)
```

## Data Validation

Checking:
- Missing values
- Duplicate records
- Data types

```python
for name, df in {
    "Fund Master": fund_master,
    "NAV History": nav_history,
    "AUM": aum,
    "SIP": sip
}.items():

    print(f"\n{name}")
    print(df.isnull().sum())
```
