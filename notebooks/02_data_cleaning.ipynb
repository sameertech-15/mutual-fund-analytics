# Data Cleaning Notebook

This notebook performs cleaning and preprocessing on all datasets.

Tasks:
- Remove duplicates
- Handle missing values
- Standardize column formats
- Validate numerical fields

```python
import pandas as pd

nav = pd.read_csv("../data/raw/02_nav_history.csv")
transactions = pd.read_csv("../data/raw/08_investor_transactions.csv")
```

```python
print("Before Cleaning")
print(nav.shape)
print(transactions.shape)
```

```python
nav.drop_duplicates(inplace=True)
transactions.drop_duplicates(inplace=True)
```

```python
nav["date"] = pd.to_datetime(nav["date"])

nav = nav.sort_values(
    ["amfi_code", "date"]
)
```

```python
nav["nav"] = nav.groupby(
    "amfi_code"
)["nav"].ffill()
```

```python
transactions = transactions[
    transactions["amount_inr"] > 0
]
```

```python
print("After Cleaning")
print(nav.shape)
print(transactions.shape)
```

## Cleaning Summary

Completed:

- Removed duplicate records
- Converted date columns
- Forward-filled NAV values
- Validated transaction amounts
- Standardized dataset structure

The cleaned datasets were exported to the `data/processed` folder and loaded into SQLite.
